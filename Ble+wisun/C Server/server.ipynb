```python
import socket
```

```python
Host = '10.1.36.11'
Port = 9090
#
# 80  - http
# 443 - https
# 21  - ftp
# 22  - ssh
# lower number reserved for common uses

host = socket.gethostbyname(socket.gethostname()) #get the hostname of the machine
print(host) 
#3 types, 192,10,172

server = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
server.bind((Host, Port)) #bind the server to the host and port

server.listen(5) #5 is the number of connections that can be queued
#5 is the unaccepted connection that we allow befor we start refusing new connections

```

```python
while True:
    communication_socket, address = server.accept() #accept the connection
    print(f"Connection from {address} has been established")
    message = communication_socket.recv(1024).decode('utf-8') #receive the message from the client
    print(f"Message from the client: {message}")
    communication_socket.send("Message received".encode('utf-8')) #send the message to the client
    communication_socket.close() #close the connection
    print(f"Connection from {address} has been closed")
    
```
